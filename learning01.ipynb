```python
# Load the CSV file
df = pd.read_csv('/kaggle/input/mall-customers/Mall_Customers.csv')
# Display the first 10 rows
df.head(10)
```

```python
# Check for duplicated rows in the DataFrame
df.duplicated().sum()
```

```python
# Display general information about the dataset
df.info()
```

```python
# Show the number of rows and columns.
df.shape
```

```python
# Group the data by 'Genre' and calculate basic statistics for the 'Age' column
df.groupby(['Genre'])['Age'].agg(['count','min','max','median','mean'])
```

```python
# Display the 5 customers with the highest Spending Scores
display(df.loc[df['Spending Score (1-100)'].nlargest(5).index])
# Display the 5 customers with the lowest Spending Scores
display(df.loc[df['Spending Score (1-100)'].nsmallest(5).index])
```

```python
# Select the customer(s) with the maximum Spending Score
# Select the customer(s) with the minimum Spending Score
score_max = df[df['Spending Score (1-100)'] == df['Spending Score (1-100)'].max()]
score_min = df[df['Spending Score (1-100)'] == df['Spending Score (1-100)'].min()]
extremes = pd.concat([score_max, score_min])# Combine both datasets
display(extremes)
```

```python
# Create a new DataFrame containing customers with high Spending Scores (≥ 71)
# Sort Spending Score in descending order
high_score = df[df['Spending Score (1-100)'] >= 71 ].sort_values(by=['Spending Score (1-100)'], ascending=[False])
display(high_score[:10])
len(high_score.index)# Count how many index values correspond to the highest scores.
```

```python
# Display the count of each gender in the high Spending Score group
high_score['Genre'].value_counts()
```

```python
# Group the genres with the highest scores and display basic statistics.
high_score_genre = high_score.groupby('Genre')['Spending Score (1-100)'].agg(['count','min','max','mean'])
display(high_score_genre)
```
